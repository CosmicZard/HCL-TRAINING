```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import plotly.express as px 

df =  pd.read_csv('Sales_Data.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df['Customer_Name'].mode()
```

```python
df['Customer_Name']=df['Customer_Name'].dropna
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df.isnull().sum()
```

```python
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df.isnull().sum()
```

```python
df['City'] = df['City'].fillna(df['City'].mode()[0])
df.isnull().sum()
```

```python
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
df.isnull().sum()
```

```python
df['Product']=df['Product'].fillna(df['Product'].mode()[0])
df.isnull().sum()
df['Quantity']=df['Quantity'].fillna(df['Quantity'].mean())
df.isnull().sum()
```

```python
df['Unit_Price_INR']=df['Unit_Price_INR'].fillna(df['Unit_Price_INR'].median())
df.isnull().sum()
df['Order_Date']=df['Order_Date'].dropna()
df.isnull().sum()
df['Sales_Channel']=df['Sales_Channel'].fillna(df['Sales_Channel'].mode()[0])
df.isnull().sum()
df['Profit_Method']=df['Profit_Method'].fillna(df['Profit_Method'].mode()[0])
df.isnull().sum()

.
```
