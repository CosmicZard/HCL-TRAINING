```python
q['Region']=q['Region'].replace('North','North America')
q['Region']=q['Region'].replace('South','South America')
q['Region'].unique()
```

```python
import matplotlib.pyplot as plt
q['Region'].value_counts().plot(kind='bar')
plt.xlabel('Region')
plt.ylabel('Count')
plt.title('Sales by Region')
plt.show()
```

```python
q.groupby(['Region', 'Product'])['Unit_Price_INR'].sum()
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [20, 10, 30, 25, 15, 5]
plt.scatter(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot Example')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

q = pd.read_csv("sales_dataset_50_examples.csv")

plt.pie(
    q['Region'].value_counts(),
    labels=q['Region'].value_counts().index,
    autopct='%1.1f%%'
)
plt.title('Sales by Region')
plt.show()
```

```python
plt.hist(q['Unit_Price_INR'], bins=10, color='blue', alpha=0.7)
plt.xlabel('Unit Price (INR)')
plt.ylabel('Frequency')
plt.title('Histogram of Unit Price')
plt.show()
```
